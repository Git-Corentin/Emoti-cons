```python
import pandas as pd, numpy as np
from gensim import corpora
from gensim.models import LdaModel, KeyedVectors
from gensim.utils import simple_preprocess
from nltk.corpus import stopwords
from sklearn.metrics.pairwise import cosine_similarity
from scipy.stats import pearsonr
from tqdm import tqdm
import re

stop = set(stopwords.words('english'))

def preprocess(text):
    if not isinstance(text, str):
        return []
    # delete URLs, mentions, hashtags, numbers, punctuation, HTML entities
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"#\w+", "", text)
    text = re.sub(r"&amp;", " ", text)     # HTML entity
    text = re.sub(r"[^A-Za-z\s]", " ", text)  # only letters / spaces
    toks = simple_preprocess(text, deacc=True)
    toks = [t for t in toks if t not in stop and len(t) > 2]
    return toks

def lda_one_topic_topk_for_emoji(df_texts, wv, num_words=8):
    # preprocess texts
    texts = [preprocess(t) for t in df_texts if isinstance(t, str)]
    dictionary = corpora.Dictionary(texts)
    dictionary.filter_extremes(no_below=5, no_above=0.5)
    
    if len(dictionary) == 0:
        return []

    corpus = [dictionary.doc2bow(t) for t in texts if len(t) > 0]

    if not corpus or all(len(doc) == 0 for doc in corpus):
        return []

    # LDA
    lda = LdaModel(
        corpus=corpus,
        id2word=dictionary,
        num_topics=1,
        random_state=42,
        iterations=100,
        passes=5,
        minimum_probability=0.0
    )

    topic = lda.show_topic(0, topn=num_words)  # [('word', weight), ...]
    return topic

emotion_names = ['anger','anticipation','disgust','fear','joy','sadness','surprise','trust']

def keywords_to_emotion_scores(topic_words, wv):
    # topic_words : list of (word, weight)
    # wv : keyedvectors supporting similarity; fallback: cosine on embeddings if using transformer
    # Return vector of length 8
    scores = np.zeros(len(emotion_names))
    for word, w in topic_words:
        # compute similarity of word to each emotion name
        s_list = []
        for i, emo in enumerate(emotion_names):
            try:
                sim = wv.similarity(word, emo)
            except Exception:
                # if OOV, try lowercased/emotion synonyms or set sim=0
                sim = 0.0
            scores[i] += sim * w
    # normalize
    if scores.sum() != 0:
        scores = scores / (np.abs(scores).sum())
    return scores

results = []
for emoji_name, texts in tqdm(df_tweets_dict.items()):
    topic = lda_one_topic_topk_for_emoji(texts, wv, num_words=8)
    pred_vec = keywords_to_emotion_scores(topic, wv)  # predicted emotion vector
    # get ground truth vector from emotag_df (match on name or normalized)
    row = emotag_df[emotag_df['normalized_name']==emoji_name]
    if row.empty:
        gt = np.zeros(len(emotion_names))
    else:
        gt = row[emotion_names].values.flatten()
        if gt.sum() != 0:
            gt = gt / gt.sum()
    # similarity / correlation
    cos = cosine_similarity(pred_vec.reshape(1,-1), gt.reshape(1,-1))[0,0]
    try:
        pearson = pearsonr(pred_vec, gt)[0]
    except Exception:
        pearson = np.nan
    results.append({'emoji':emoji_name, 'lda_keywords':[w for w,_ in topic],
                    'pred_vec':pred_vec, 'gt_vec':gt, 'cosine':cos, 'pearson':pearson})
res_df = pd.DataFrame(results)
```

```python
import pandas as pd
import numpy as np
from tqdm import tqdm
import sys
sys.modules["torchvision"] = None  # workaround for import error
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import re
from gensim.utils import simple_preprocess
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer




# Keep only those emojis
common_emojis = set(df_tweets_dict.keys()) & set(emotag_df["normalized_name"])
df_tweets_dict = {k: v for k, v in df_tweets_dict.items() if k in common_emojis}
emotag_df = emotag_df[emotag_df["normalized_name"].isin(common_emojis)].reset_index(drop=True)

print(f"{len(common_emojis)} emojis kept (present in EmoTag).")

stop = set(stopwords.words("english"))
extra_stop = {"amp", "rt", "im", "u", "ur", "dont", "cant", "youre", "ive", "lol", "omg"}
stop |= extra_stop

def preprocess(text):
    if not isinstance(text, str):
        return []
    # delete URLs, mentions, hashtags, numbers, punctuation, HTML entities
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    text = re.sub(r"@\w+|#\w+", "", text)
    text = re.sub(r"&amp;", " ", text)     # HTML entité
    text = re.sub(r"[^A-Za-z\s]", " ", text)  # garde que lettres / espaces
    toks = simple_preprocess(text, deacc=True)
    toks = [t for t in toks if t not in stop and len(t) > 2]
    return toks

# Load sentence-transformer model
embedder = SentenceTransformer("all-MiniLM-L6-v2")

def bertopic_for_emoji(texts, nr_topics="auto", top_n_words=8, sample_size=3000):
    """Compute BERTopic on tweets of one emoji (downsampled)."""
    # Downsample to speed up (3k tweets max)
    if len(texts) > sample_size:
        texts = texts.sample(sample_size, random_state=42)
    else:
        texts = texts.dropna()

    texts = [" ".join(preprocess(t)) for t in texts if isinstance(t, str) and len(t) > 0]
    texts = [t for t in texts if len(t.split()) > 2]  # we ignore too-short tweets

    if len(texts) < 20:
        return None, None, None, None, None

    embeddings = embedder.encode(texts, show_progress_bar=False, batch_size=64)
    vectorizer_model = CountVectorizer(stop_words="english")

    topic_model = BERTopic(
        nr_topics=nr_topics,
        min_topic_size=50,
        verbose=False,
        calculate_probabilities=True,
        vectorizer_model=vectorizer_model,
    )

    topics, probs = topic_model.fit_transform(texts, embeddings)
    freq = pd.Series(topics).value_counts()
    top_topic = int(freq.index[0])
    words = topic_model.get_topic(top_topic)[:top_n_words] if topic_model.get_topic(top_topic) else []
    return topic_model, top_topic, words, topics, probs

emotion_names = ['anger','anticipation','disgust','fear','joy','sadness','surprise','trust']

def keywords_to_emotion_scores(topic_words, wv):
    scores = np.zeros(len(emotion_names))
    for word, w in topic_words:
        for i, emo in enumerate(emotion_names):
            try:
                sim = wv.similarity(word, emo)
            except Exception:
                sim = 0.0
            scores[i] += sim * w
    if scores.sum() != 0:
        scores = scores / np.abs(scores).sum()
    return scores

bert_results = []

for emoji_name, texts in tqdm(df_tweets_dict.items()):
    tm, top_topic, words, topics, probs = bertopic_for_emoji(texts)
    if not words:
        continue

    topic_words = [(w, score) for w, score in words]
    pred_vec = keywords_to_emotion_scores(topic_words, wv)

    row = emotag_df[emotag_df["normalized_name"] == emoji_name]
    gt = row[emotion_names].values.flatten() if not row.empty else np.zeros(len(emotion_names))
    if gt.sum() != 0:
        gt = gt / gt.sum()

    cos = cosine_similarity(pred_vec.reshape(1, -1), gt.reshape(1, -1))[0, 0]
    bert_results.append({
        "emoji": emoji_name,
        "topic_words": [w for w, _ in topic_words],
        "pred_vec": pred_vec,
        "gt_vec": gt,
        "cosine": cos
    })

bert_df = pd.DataFrame(bert_results)
print(f"BERTopic completed for {len(bert_df)} emojis.")

bert_df.to_csv("bertopic_results.csv", index=False)
display(bert_df.sort_values("cosine", ascending=False))
```

```python
import sys
import os
import torch
import numpy as np
import pandas as pd
from tqdm import tqdm
import json
import emoji as emoji_lib

from torchmoji.sentence_tokenizer import SentenceTokenizer
from torchmoji.model_def import torchmoji_emojis
from torchmoji.global_variables import PRETRAINED_PATH, VOCAB_PATH

print("Loading TorchMoji model and vocabulary...")
with open(VOCAB_PATH, 'r') as f:
    vocabulary = json.load(f)

maxlen = 30
st = SentenceTokenizer(vocabulary, maxlen)
model = torchmoji_emojis(PRETRAINED_PATH)
model.eval()
print("Model loaded.")

EMOJIS = ":joy: :unamused: :weary: :sob: :heart_eyes: \
:pensive: :ok_hand: :blush: :heart: :smirk: \
:grin: :notes: :flushed: :100: :sleeping: \
:relieved: :relaxed: :raised_hands: :two_hearts: :expressionless: \
:sweat_smile: :pray: :confused: :kissing_heart: :heartbeat: \
:neutral_face: :information_desk_person: :disappointed: :see_no_evil: :tired_face: \
:v: :sunglasses: :rage: :thumbsup: :cry: \
:sleepy: :yum: :triumph: :hand: :mask: \
:clap: :eyes: :gun: :persevere: :smiling_imp: \
:sweat: :broken_heart: :yellow_heart: :musical_note: :speak_no_evil: \
:wink: :skull: :confounded: :smile: :stuck_out_tongue_winking_eye: \
:angry: :no_good: :muscle: :facepunch: :purple_heart: \
:sparkling_heart: :blue_heart: :grimacing: :sparkles:".split(' ')

emoji_name_map = {}
for idx, alias in enumerate(EMOJIS):
    try:
        uni = emoji_lib.emojize(alias, language='alias')
        name = emoji_lib.demojize(uni).strip(':').replace(' ', '_')
        emoji_name_map[name] = idx
    except:
        pass

print(f"{len(emoji_name_map)} emojis mapped to names.")

def predict_emoji_for_texts(texts, model, st, top_k=5, batch_size=256):
    preds = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i + batch_size]
        tokenized, _, _ = st.tokenize_sentences(batch)
        tokenized = torch.tensor(tokenized.astype(np.int64))

        with torch.no_grad():
            logits = model(tokenized)
            probs = torch.softmax(logits, dim=1).cpu().numpy()

        # indices of top-k predictions
        top_preds = np.argsort(probs, axis=1)[:, -top_k:][:, ::-1]
        preds.extend(top_preds.tolist())

    return preds

results = []
topk_values = [1, 2, 3, 4, 5]

for emoji_name, texts in tqdm(df_tweets_dict.items(), desc="Evaluating emojis"):
    if emoji_name not in emoji_name_map:
        continue

    if len(texts) > 5000:
        texts = texts.sample(5000, random_state=42)

    target_id = emoji_name_map[emoji_name]
    emoji_res = {"emoji": emoji_name}

    for k in topk_values:
        preds = predict_emoji_for_texts(list(texts), model, st, top_k=k)
        match_prop = np.mean([1 if target_id in p else 0 for p in preds])
        emoji_res[f"top@{k}"] = match_prop

    results.append(emoji_res)

deepmoji_df = pd.DataFrame(results)
deepmoji_df = deepmoji_df.sort_values('top@1', ascending=False).reset_index(drop=True)

print("\nDeepMoji evaluation completed.\n")
display(deepmoji_df)

print("Top-10 emojis by Top@1 accuracy:\n")
print(deepmoji_df.to_string(index=False))
```

```python
from collections import Counter
from nltk.corpus import wordnet as wn, stopwords
from gensim.utils import simple_preprocess
import re

# Stopwords classiques + spécifiques à Twitter
stop = set(stopwords.words('english'))
extra_stop = {
    'co','https','http','amp','rt','im','youre','ive','get','got','one','us',
    'see','know','really','good','time','like','go','much','would','today',
    'love','thank','thanks','day','great','make','well','back','even','still',
    'think','want', 'easter', 'egg', 'thats', 'always', 'dont', 'literally', 'yall',
    'gonna', 'need', 'way', 'cant', 'also', 'going', 'right', 'didnt', 'never', 'people',
    'man', 'bro', 'say', 'said', 'hes', 'omg', 'look', 'stop', 'thing', 'ill'
}
stop |= extra_stop

def preprocess_text(text):
    text = re.sub(r'http\S+|www\S+|@\w+|#\w+', '', str(text))  # remove URLs, mentions, hashtags
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # keep letters only
    return [w for w in simple_preprocess(text) if w not in stop and len(w) > 2]

def antonyms_of_word(w):
    ants = set()
    for syn in wn.synsets(w):
        for lemma in syn.lemmas():
            for ant in lemma.antonyms():
                ants.add(ant.name().lower().replace('_',' '))
    return ants

def top_n_terms_with_antonyms(texts, n=20, pool=200):
    toks = []
    for t in texts:
        toks.extend(preprocess_text(t))
    freq_words = [w for w, _ in Counter(toks).most_common(pool)]
    valid = [w for w in freq_words if len(antonyms_of_word(w)) > 1]
    return valid[:n]

def antonym_overlap(H1, H2):
    H1 = [h.lower() for h in H1]
    H2s = set(h.lower() for h in H2)
    matched = []
    for w in H1:
        ants = antonyms_of_word(w)
        if len(ants & H2s) > 0:
            matched.append((w, list(ants & H2s)))
    prop = len(matched) / len(H1) if len(H1) > 0 else 0
    return prop, matched

# 🔹 Exemple : sadness vs happiness
print("Available emoji keys:", list(df_tweets_dict.keys())[:10])

H_sad = top_n_terms_with_antonyms(df_tweets_dict['loudly_crying_face'], n=20)
H_joy = top_n_terms_with_antonyms(df_tweets_dict['smiling_face'], n=20)

print("\nFiltered sadness terms (with antonyms):", H_sad)
print("Filtered joy terms (with antonyms):", H_joy)

prop_sad_in_joy, matched_sad = antonym_overlap(H_sad, H_joy)
prop_joy_in_sad, matched_joy = antonym_overlap(H_joy, H_sad)

print("\nProportion of sadness words that have antonyms in joy:", round(prop_sad_in_joy,3))
print("Proportion of joy words that have antonyms in sadness:", round(prop_joy_in_sad,3))
print("\nExample antonym pairs (sad→joy):", matched_sad)
print("Example antonym pairs (joy→sad):", matched_joy)
```
